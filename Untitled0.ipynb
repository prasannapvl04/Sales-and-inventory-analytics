```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# 1. LOAD DATA
# =====================================================

df = pd.read_csv("/content/hero_honda_bike_orders_dataset.csv")

df['Price (INR)'] = (
    df['Price (INR)']
    .astype(str)
    .str.replace('?', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(float)
)

print(df.head())
print(df.shape)
print(df.isnull().sum())


# =====================================================
# 2. PRICE ANALYSIS
# =====================================================

price = df["Price (INR)"].to_numpy()

print("\nPRICE ANALYSIS")
print("Mean:", np.mean(price))
print("Median:", np.median(price))
print("Std:", np.std(price))
print("Variance:", np.var(price))
print("Min:", np.min(price))
print("Max:", np.max(price))


# =====================================================
# 3. BIKE TYPE ANALYSIS
# =====================================================

bike_type = df.groupby("Bike Type")["Price (INR)"].agg(
    ["count", "sum", "mean"]
)

print("\nBIKE TYPE ANALYSIS")
print(bike_type)

print("Highest Revenue Bike Type:",
      bike_type["sum"].idxmax())


# =====================================================
# 4. PERCENTILE & PREMIUM ORDERS
# =====================================================

p25, p50, p75, p90 = np.percentile(
    price, [25, 50, 75, 90]
)

print("\nPERCENTILES")
print("25%:", p25)
print("50%:", p50)
print("75%:", p75)
print("90%:", p90)

premium = df[price > p90]

print("Premium Orders:", len(premium))


# =====================================================
# 5. PRICE RANGE
# =====================================================

bins = [0, 70000, 100000, 150000, np.inf]

counts, _ = np.histogram(price, bins=bins)

print("\nPRICE RANGE COUNTS")
print("<70K:", counts[0])
print("70K-1L:", counts[1])
print("1L-1.5L:", counts[2])
print(">1.5L:", counts[3])


# =====================================================
# 6. DELIVERY ANALYSIS
# =====================================================

date_cols = [
    "Order Date",
    "Expected Delivery Date",
    "Received Date"
]

for col in date_cols:

    df[col] = pd.to_datetime(df[col], errors='coerce')


df_delivery = df.dropna(subset=["Received Date", "Expected Delivery Date"])

df_delivery["Delivery Delay"] = (
    (df_delivery["Received Date"] - df_delivery["Expected Delivery Date"]).dt.days
)

delay = df_delivery["Delivery Delay"].to_numpy()

print("\nDELIVERY ANALYSIS")
print("Average Delay:", np.mean(delay), "days")
print("Late:", np.sum(delay > 0))
print("Early:", np.sum(delay < 0))
print("On-Time:", np.sum(delay == 0))

print("\nOrders with delay > 5 days:")
print(df_delivery[delay > 5])


# =====================================================
# 7. CANCELLED ORDERS
# =====================================================

cancelled = df[
    df["Order Status"].str.lower() == "cancelled"
]

print("\nCANCELLED ORDERS")
print("Count:", len(cancelled))
print(
    "Percentage:",
    len(cancelled) / len(df) * 100
)


# =====================================================
# 8. PRODUCT STATUS
# =====================================================

status = df.groupby(
    "Current Product Status"
)["Price (INR)"].agg(["count", "sum"])

status["Revenue %"] = (
    status["sum"] /
    df["Price (INR)"].sum() * 100
)

print("\nPRODUCT STATUS")
print(status)


# =====================================================
# 9. CUMULATIVE REVENUE
# =====================================================

df = df.sort_values("Order Date")
df["Revenue"] = df["Price (INR)"]
df["Cumulative Revenue"] = np.cumsum(
    df["Revenue"].to_numpy()
)
print("\nTOTAL REVENUE:",
      np.sum(df["Revenue"].to_numpy()))
plt.plot(
    df["Order Date"],
    df["Cumulative Revenue"]
)
plt.title("Cumulative Revenue")
plt.xlabel("Order Date")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()

# =====================================================
# 10. CORRELATION
# =====================================================

if not df_delivery.empty:
    correlation = np.corrcoef(
        df_delivery["Price (INR)"],
        df_delivery["Delivery Delay"]
    )[0, 1]
else:
    correlation = np.nan
print("\nPRICE vs DELIVERY DELAY")
print("Correlation:", correlation)


# =====================================================
# FINAL SUMMARY
# =====================================================

print("\n========== FINAL RESULTS ==========")
print("Total Orders:", len(df))
print("Total Revenue:", df["Revenue"].sum())
print("Average Price:", np.mean(price))
print("Premium Orders:", len(premium))
print("Cancelled Orders:", len(cancelled))
print("Late Orders:", np.sum(delay > 0))
print(
    "Top Bike Type:",
    bike_type["sum"].idxmax()
)
print(
    "Price-Delay Correlation:",
    correlation
)
```
